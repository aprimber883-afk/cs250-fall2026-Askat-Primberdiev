# Lockout Quote
**Askat Primberdiev** · CS 250 Final Project · Fall 2026

**Pitch:** A simple price calculator and weekly tracker for mobile locksmiths.

I work as a mobile car locksmith in San Jose. When a customer calls, I need to give a price fast. At the end of the week I want to see how much I made and which jobs pay the most.

```python
# Imports
import matplotlib.pyplot as plt
```

```python
# My prices (in dollars)
PRICES = {
    "car lockout": 85,
    "house lockout": 95,
    "key cutting": 60,
    "key programming": 150,
}

# My jobs from one week
WEEK = [
    {"type": "car lockout", "price": 125},
    {"type": "key programming", "price": 210},
    {"type": "house lockout", "price": 115},
    {"type": "car lockout", "price": 165},
    {"type": "key cutting", "price": 80},
    {"type": "car lockout", "price": 190},
    {"type": "key programming", "price": 245},
    {"type": "house lockout", "price": 150},
]
```

```python
# My functions

def quote_price(job_type, miles, night, weekend):
    """Return the price of a job: base price + $2 per mile + $40 at night + $25 on weekend."""
    price = PRICES[job_type] + miles * 2
    if night:
        price = price + 40
    if weekend:
        price = price + 25
    return price


def total_money(jobs):
    """Return the total money from a list of jobs."""
    total = 0
    for job in jobs:
        total = total + job["price"]
    return total


def average_price(jobs):
    """Return the average price of one job."""
    return total_money(jobs) / len(jobs)


def money_by_type(jobs):
    """Return a dictionary with the total money for each job type."""
    totals = {}
    for job in jobs:
        if job["type"] in totals:
            totals[job["type"]] = totals[job["type"]] + job["price"]
        else:
            totals[job["type"]] = job["price"]
    return totals
```

```python
# Quick tests with answers I know
print(quote_price("car lockout", 10, True, False))   # 85 + 20 + 40 = 145
print(total_money(WEEK))                             # 1280
print(average_price(WEEK))                           # 160.0
```

```python
# main(): runs my program

def main():
    """Ask about a new job and print the price, then print my week."""
    print("=== Lockout Quote ===")
    job_type = input("Job type (car lockout, house lockout, key cutting, key programming): ")
    miles = int(input("Miles to drive: "))
    night = input("Night? (y/n): ") == "y"
    weekend = input("Weekend? (y/n): ") == "y"

    if job_type in PRICES:
        print("Price:", quote_price(job_type, miles, night, weekend), "dollars")
    else:
        print("I don't have this job type")

    print()
    print("=== My week ===")
    print("Jobs:", len(WEEK))
    print("Total:", total_money(WEEK), "dollars")
    print("Average per job:", average_price(WEEK), "dollars")
    print("By job type:", money_by_type(WEEK))

main()
```

```python
# Results: graph of money by job type
totals = money_by_type(WEEK)
plt.bar(totals.keys(), totals.values(), color="orange")
plt.title("Money by job type (one week)")
plt.ylabel("Dollars")
plt.show()
```
